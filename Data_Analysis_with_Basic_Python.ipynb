```python

#  ASSIGNMENT 1 – Data Analysis with Basic Python

# Question 1:

# Ek program likhein jo user se 15 numbers input le aur:
# Maximum aur minimum manually find karein (built-in functions use na karein).
# Average calculate karein.
# Total even aur odd numbers count karein.
# Batayein kitne numbers average se zyada hain.
# Ek separate function likhein jo list reverse kare without using reverse().

# 📌Condition:

# Loops aur conditions ka use zaroor karein.
# Code modular hona chahiye (functions use karein).
# ============================================


# ============================================
# user input of 15 num
# ============================================
def user_input():
  number_list=[]
  for i in range(5):
     num=int(input(f"Enter number:"))
     number_list.append(num)
  return number_list

# ============================================
# max and min finding manually
# ============================================
def find_max_min(return_list):
  max_val=return_list[0]
  min_val=return_list[0]
  for i in return_list:
    if i >max_val:
      max_val=i
    elif i< min_val:
      min_val=i

  return max_val,min_val

# ============================================
# Average calculate karein.
# ============================================
def calculate_average(return_list):
  total=sum(return_list)
  average=total/len(return_list)
  return average

# ============================================
# Total even aur odd numbers count karein.
# ============================================
def count_even_odd(return_list):
  even_count=0
  odd_count=0
  for i in return_list:
    if i % 2==0:
      even_count+=1
    else:
      odd_count+=1
  return even_count,odd_count

# ============================================
# Batayein kitne numbers average se zyada hain.
# ============================================
def count_above_average(return_list,average):
  count=0
  for i in return_list:
    if i > average:
      count+=1
  return count

# ============================================
# Ek separate function likhein jo list reverse kare without using reverse().
# ============================================
def reverse_list(return_list):
  reversed_list=[]
  for i in range(len(return_list)-1,-1,-1):
    reversed_list.append(return_list[i])
  return reversed_list

# ============================================
#CALLING FUNCTION:
# ============================================
print("="*40)
print("INPUT TAKING BY USER:")
print("="*40)
return_list=user_input()
print(return_list)
print("\n")

print("="*40)
print("MAXIMUM OR MINIMUM VALUE OF NUMBERS:")
print("="*40)
maximum, minimum=find_max_min(return_list)
print("Maximum value is:", maximum)
print("Minimum value is:", minimum)
print("\n")

print("="*40)
print(" EVEN OR ODD OF NUMBERS:")
print("="*40)
result=count_even_odd(return_list)
print("Even:", result[0])
print("Odd:", result[1])
print("\n")


print("="*40)
print("AVERAGE OF NUMBERS:")
print("="*40)
average=calculate_average(return_list)
print("Average value is:", average)
print("\n")

print("="*40)
print("NUMBER ABOVE THEN AVERAGE:")
print("="*40)
result=count_above_average(return_list,average)
print("Numbers above average:", result)
print("\n")

print("="*40)
print("REVERSED THE LIST:")
print("="*40)
result=reverse_list(return_list)
print(f"Original: {return_list}")
print(f"Reversed: {result}")
```

```python
# ASSIGNMENT 2 – Library Book Management
#  Question 2:
# Ek simple library system banayein jisme:
# Books ki list create karein.
# User ko option mile:
# Add book
# Remove book
# Search book
# Display all books
# Agar book exist nahi karti to proper message show karein.
# Ek function likhein jo total books count kare.

# 📌 Conditions:

# Lists aur functions ka proper use karein.
# Menu-driven program ho.
# ============================================

# ============================================
#Ek simple library system banayein jisme:
# Books ki list create karein.
# ============================================
library_books = []

# ============================================
# Add Book Function
# ============================================
def add_book(library_books):
    book = input("Enter book name: ")
    if book in library_books:
        print(" Book already exists!")
    else:
        library_books.append(book)
        print("Book added!")

# ============================================
# Remove Book Function
# ============================================
def remove_book(library_books):
    book = input("Enter book name to remove: ")
    if book in library_books:
        library_books.remove(book)
        print(" Book removed!")
    else:
        print(" Book not found!")

# ============================================
# Search Book Function
# ============================================
def search_book(library_books):
    book = input("Enter book name to search: ")
    if book in library_books:
        print(" Book found!")
    else:
        print(" Book not found!")

# ============================================
# Display Books Function
# ============================================
def display_books(library_books):
    if len(library_books) == 0:
        print("Library is empty!")
    else:
        print("\n=== All Books ===")
        for i, book in enumerate(library_books, 1):
            print(f"{i}. {book}")
        print(f"Total Books: {len(library_books)}")

# ============================================
#  Main Menu (While Loop)
# ============================================
while True:
    print("\n" + "="*40)
    print(" LIBRARY MANAGEMENT SYSTEM")
    print("="*40)
    print("1. Add Book")
    print("2. Remove Book")
    print("3. Search Book")
    print("4. Display All Books")
    print("5. Exit")
    print("="*40)

    choice = input("Enter your choice (1-5): ")

    if choice == '1':
        add_book(library_books)
    elif choice == '2':
        remove_book(library_books)
    elif choice == '3':
        search_book(library_books)
    elif choice == '4':
        display_books(library_books)
    elif choice == '5':
        print(" Thank you! Goodbye!")
        break
    else:
        print(" Invalid choice! Please enter 1-5.")
```

```python
# ASSIGNMENT 3 – Recursion Mastery

# Question 3:

# Neeche diye gaye problems recursively solve karein:
# Factorial function likhein.
# Fibonacci sequence ka nth term find karein.
# Ek recursive function likhein jo list ka sum calculate kare.
# Ek function likhein jo string palindrome hai ya nahi check kare (recursively).
# Ek power function likhein jo a^b calculate kare without using **.

# 📌 Condition:

# Har function mein recursion ka proper base case hona chahiye.
# Loop use nahi karna.
# ============================================


# ============================================
# Factorial function likhein.
# ============================================

def factorial_normal(n):
    if n == 0 or n == 1:
        return 1

    result = 1
    #  Loop use karein
    for i in range(2, n + 1):
        result = result * i

    return result


# ============================================
# Fibonacci sequence ka nth term find karein.
# ============================================
def fibonacci_normal(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    a, b = 0, 1
    for i in range(2, n + 1):
        a, b = b, a + b

    return b


# ============================================
# Ek recursive function likhein jo list ka sum calculate kare.
# ============================================
def sum_list_normal(numbers):
    if len(numbers) == 0:
        return 0

    total = 0
    #  Loop use karein
    for num in numbers:
        total = total + num

    return total


# ============================================
# Ek function likhein jo string palindrome hai ya nahi check kare (recursively).
# ============================================
def is_palindrome_normal(s):
    if len(s) <= 1:
        return True
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left = left + 1
        right = right - 1

    return True


# ============================================
# Ek power function likhein jo a^b calculate kare without using **.
# ============================================
def power_normal(a, b):
    if b == 0:
        return 1
    result = 1
    for i in range(b):
        result = result * a

    return result


# ============================================
# CALLING FUNCTION:
# ============================================
def main():
    print("="*40)
    print("FACTORIAL FUNCTION:")
    print("="*40)
    for i in range(6):
        print(f"{i}! = {factorial_normal(i)}")
    print("\n")

    print("="*40)
    print("NTH TERM OF FIBONACCI SEQUENCE:")
    print("="*40)
    for i in range(10):
        print(f"Fibonacci({i}) = {fibonacci_normal(i)}")
    print("\n")

    print("="*40)
    print("SUM OF LIST:")
    print("="*40)
    test_list = [1, 2, 3, 4, 5]
    print(f"List: {test_list}")
    print(f"Sum: {sum_list_normal(test_list)}")
    print("\n")

    print("="*40)
    print("CHECKING STRING PALINDROME OF LIST:")
    print("="*40)
    test_strings = ["madam", "hello", "racecar", "python"]
    for s in test_strings:
        result = " Palindrome" if is_palindrome_normal(s) else " Not Palindrome"
        print(f"'{s}' → {result}")
    print("\n")

    print("="*40)
    print("CALCULATING POWER OF NUMBER:")
    print("="*40)
    test_cases = [(2, 3), (5, 2), (3, 4), (10, 0)]
    for a, b in test_cases:
        print(f"{a}^{b} = {power_normal(a, b)}")
    print("\n")

if __name__ == "__main__":
    main()
```

```python
# 📝 ASSIGNMENT 4 – Bank Account System (Encapsulation)

# Question 4:

# Ek class BankAccount design karein jisme:
# Attributes:
# account_holder
# __balance (private variable)
# Methods:
# deposit(amount)
# withdraw(amount)
# check_balance()
# transfer(other_account, amount)
# Program mein:
# Do accounts create karein.
# Money transfer simulate karein.
# Withdraw balance se zyada na ho.

# 📌 Conditions:
# Balance private hona chahiye.
# Getter aur setter implement karein.
# Transaction history maintain karein (list use karein).


# ============================================
# BANK ACCOUNT CLASS (Encapsulation):
# ============================================
class BankAccount:
    def __init__(self,account_holder, initial_balance=0):
        self.account_holder = account_holder
        self.__balance = initial_balance
        self.__transaction_history = []

    # ============================================
    #  GETTER METHOD():
    # ============================================
    def get_balance(self):
        return self.__balance

    # ============================================
    # SETTER METHOD():
    # ============================================
    def set_balance(self, new_balance):
        if new_balance >= 0:
            self.__balance = new_balance
            self.__add_transaction("Balance Updated", new_balance)
        else:
            print(" Balance negative nahi ho sakta!")

    # ============================================
    # deposit(amount)
    # ============================================
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__add_transaction("Deposit", amount)
            print(f"{amount} deposited successfully!")
            print(f"New Balance: {self.__balance}")
        else:
            print("Deposit amount positive hona chahiye!")

    # ============================================
    # withdraw(amount)
    # ============================================
    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                self.__add_transaction("Withdraw", -amount)
                print(f"{amount} withdrawn successfully!")
                print(f"New Balance: {self.__balance}")
            else:
                print(f" Insufficient Balance! Available: {self.__balance}")
        else:
            print(" Withdraw amount positive hona chahiye!")

    # ============================================
    # transfer(other_account, amount)
    # ============================================
    def transfer(self, other_account, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                self.__add_transaction(f"Transfer to {other_account.account_holder}", -amount)

                other_account.__balance += amount
                other_account.__add_transaction(f"Transfer from {self.account_holder}", amount)

                print(f"{amount} transferred to {other_account.account_holder}!")
                print(f"Your Balance: {self.__balance}")
                print(f"{other_account.account_holder}'s Balance: {other_account.__balance}")
            else:
                print(f" Insufficient Balance for Transfer! Available: {self.__balance}")
        else:
            print(" Transfer amount positive hona chahiye!")

    # ============================================
    # check_balance()
    # ============================================
    def check_balance(self):
        print(f"\n Account Holder: {self.account_holder}")
        print(f" Current Balance: {self.__balance}")
        print(f" Total Transactions: {len(self.__transaction_history)}")

    # ============================================
    # TRANSACTION HISTORY
    # ============================================
    def get_transaction_history(self):
        print(f"\n Transaction History for {self.account_holder}:")
        print("="*40)
        if len(self.__transaction_history) == 0:
            print("No transactions yet!")
        else:
            for i, transaction in enumerate(self.__transaction_history, 1):
                print(f"{i}. {transaction}")
        print("="*40)

    # ============================================
    # PRIVATE METHOD (Add Transaction)
    # ============================================
    def __add_transaction(self, description, amount):
        self.__transaction_history.append(f"{description}: {amount}")


# ============================================
# MAIN PROGRAM
# ============================================
def main():
    print("="*60)
    print("BANK ACCOUNT SYSTEM - ASSIGNMENT 4")
    print("="*60)
    print("\n")

    print("="*40)
    print("CREATING ACCOUNTS:")
    print("="*40)
    print("\n Creating Two Bank Accounts...")
    account1 = BankAccount("UROOJ KHAN", 2000)
    account2 = BankAccount(" MISS MUSKAN ", 3000)
    print("\n")

    print("="*40)
    print("CHECKING INITIAL BALANCE:")
    print("="*40)
    print("\n Initial Balance Check:")
    account1.check_balance()
    account2.check_balance()
    print("\n")

    print("="*40)
    print("MONEY DEPOSITIG:")
    print("="*40)
    print("\n Depositing Money...")
    account1.deposit(1000)
    account2.deposit(1500)
    print("\n")

    print("="*40)
    print("WITHDRAW MONEY:")
    print("="*40)
    print("\n Withdrawing Money...")
    account1.withdraw(500)
    account2.withdraw(600)
    print("\n")

    print("="*40)
    print("WITHDRAW MORE THAN BALANCE:")
    print("="*40)
    print("\n Trying to withdraw more than balance...")
    account1.withdraw(3000)
    print("\n")

    print("="*40)
    print("TRANSFER MONEY:")
    print("="*40)
    print("\nTransferring Money...")
    account1.transfer(account2, 700)
    print("\n")

    print("="*40)
    print("CHECKING TOTAL BALANCE:")
    print("="*40)
    print("\nFinal Balance Check:")
    account1.check_balance()
    account2.check_balance()
    print("\n")

    print("="*40)
    print("CHECKING TRANSACTION HISTORY:")
    print("="*40)
    print("\nTransaction History:")
    account1.get_transaction_history()
    account2.get_transaction_history()
    print("\n")

    print("\n" + "="*40)
    print(" BANK ACCOUNT SYSTEM - COMPLETED SUCCESSFULLY!")
    print("="*40)

#running program
if __name__ == "__main__":
    main()
```

```python
# ASSIGNMENT 5 – Inheritance & Polymorphism

# Question 5:

# Ek base class Person banayein jisme:
# Attributes:
# name
# age
# Method:
# get_details()
# Do child classes banayein:
# Student (additional attribute: GPA)
# Teacher (additional attribute: salary)
# Tasks:
# get_details() method override karein.
# Ek list mein mixed objects store karein.
# Loop chala kar polymorphism demonstrate karein.

# 📌 Condition:

# Super() ka use karein.
# Method overriding clear hona chahiye.
# ============================================================

# ============================================================
# BASE CLASS - Person
# ============================================================
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_details(self):
        return f"Name: {self.name}, Age: {self.age}"

# ============================================================
#  CHILD CLASS 1 - Student (inherits from Person)
# ============================================================
class Student(Person):
    def __init__(self, name, age, GPA):
        super().__init__(name, GPA)  # Parent class ka constructor call
        self.GPA = GPA

    def get_details(self):
        parent_info = super().get_details()
        return f"{parent_info}, GPA: {self.GPA}"

# ============================================================
#  CHILD CLASS 2 - Teacher (inherits from Person)
# ============================================================
class Teacher(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def get_details(self):
        parent_info = super().get_details()
        return f"{parent_info}, Salary: {self.salary}"

# ============================================================
#  POLYMORPHISM - Mixed objects in one list
# ============================================================
people = [
    Student("UROOJ", 20, 3.8),
    Teacher("MISS MUSKAN", 28, 50000),
    Student("SARA", 19, 3.5),
    Teacher("Miss MANAHIL", 30, 45000)
]

# ============================================================
# Loop chalakar polymorphism demonstrate
# ============================================================
print("=== POLYMORPHISM DEMO===")
for person in people:
    print(person.get_details())
    print("-" * 30)
```

```python
# **PROJECT 1 – Simple Data Analysis Dashboard (Without Libraries)
# Problem Statement**

# Question 1:

# Ek “Mini Data Analysis Dashboard” banayein jo numerical dataset ko analyze kare.
# Dataset manually define karein (example: sales data of 12 months).

#  Project Requirements:
# Ek class DataAnalyzer banayein jisme:
# data (list of numbers)
# Methods implement karein:
# calculate_mean()
# calculate_median()
# calculate_max()
# calculate_min()
# count_above_average()
# find_trend() (increasing, decreasing ya mixed)
# Ek menu-driven interface banayein jisme:
# User analysis option choose kare.
# Har option ka result show ho.
#  Bonus:
# User new number add kar sake.
# Data update hone ke baad analytics dobara run kare.



# ======================================================
# PROJECT 1 – Simple Data Analysis Dashboard (Without Libraries)
# ======================================================

class DataAnalyzer:
    def __init__(self, data):
        self.data = data
        print("DataAnalyzer initialized successfully!\n")

    # ======================================================
    # calculate_mean()
    # ======================================================
    def calculate_mean(self):
        return sum(self.data) / len(self.data)

    # ======================================================
    # calculate_median()
    # ======================================================
    def calculate_median(self):
        sorted_data = sorted(self.data)
        n = len(sorted_data)
        if n % 2 == 0:
            return (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
        else:
            return sorted_data[n//2]

    # ======================================================
    # calculate_max()
    # ======================================================
    def calculate_max(self):
        return max(self.data)

    # ======================================================
    # calculate_min()
    # ======================================================
    def calculate_min(self):
        return min(self.data)

    # ======================================================
    # count_above_average()
    # ======================================================
    def count_above_average(self):
        avg = self.calculate_mean()
        count = 0
        for num in self.data:
            if num > avg:
                count += 1
        return count

    # ======================================================
    # find_trend() (increasing, decreasing ya mixed)
    # ======================================================
    def find_trend(self):
        increasing = 0
        decreasing = 0

        for i in range(1, len(self.data)):
            if self.data[i] > self.data[i-1]:
                increasing += 1
            elif self.data[i] < self.data[i-1]:
                decreasing += 1

        if increasing > 0 and decreasing == 0:
            return "Increasing"
        elif decreasing > 0 and increasing == 0:
            return "Decreasing"
        else:
            return "Mixed"

    # ======================================================
    # User new number add kar sake (BONUS)
    # ======================================================
    def add_data(self, new_value):
        self.data.append(new_value)
        print(f"New value {new_value} added! Total data points: {len(self.data)}")

# ======================================================
# SAMPLE SALES DATA (12 months)
# ======================================================
sales_data = [2000, 2200, 1800, 2500, 2300, 2700, 3000, 2800, 3200, 3500, 3400, 3800]

# ======================================================
# Create analyzer object
# ======================================================
analyzer = DataAnalyzer(sales_data)

def show_menu():
    print("\n=== DATA ANALYSIS DASHBOARD ===")
    print("1. Show Mean (Average)")
    print("2. Show Median")
    print("3. Show Maximum")
    print("4. Show Minimum")
    print("5. Count Above Average")
    print("6. Find Trend")
    print("7. Add New Data")
    print("8. Show Current Data")
    print("0. Exit")
    print("=" * 40)

# ======================================================
# main program
# ======================================================
def main():
    while True:
        show_menu()
        choice = input("Enter choice (0-8): ")

        if choice == '1':
            mean = analyzer.calculate_mean()
            print(f"\nMean (Average): {mean:.2f}\n")

        elif choice == '2':
            median = analyzer.calculate_median()
            print(f"\nMedian: {median:.2f}\n")

        elif choice == '3':
            max_val = analyzer.calculate_max()
            print(f"\nMaximum: {max_val}\n")

        elif choice == '4':
            min_val = analyzer.calculate_min()
            print(f"\nMinimum: {min_val}\n")

        elif choice == '5':
            above_avg = analyzer.count_above_average()
            total = len(analyzer.data)
            print(f"\nNumbers above average: {above_avg} out of {total}\n")

        elif choice == '6':
            trend = analyzer.find_trend()
            print(f"\nTrend: {trend}\n")

        elif choice == '7':
            try:
                new_value = float(input("\nEnter new sales value: "))
                analyzer.add_data(new_value)
            except ValueError:
                print("Invalid number! Try again.\n")

        elif choice == '8':
            print(f"\nCurrent Data: {analyzer.data}")
            print(f"Total data points: {len(analyzer.data)}\n")

        elif choice == '0':
            print("\nThank you for using Data Analysis Dashboard!")
            break

        else:
            print("\nInvalid choice! Try again.\n")

# ======================================================
# Start program
# ======================================================
if __name__ == "__main__":
    print("Welcome to Data Analysis Dashboard!")
    print(f"Loaded {len(sales_data)} months of sales data")
    main()
```
