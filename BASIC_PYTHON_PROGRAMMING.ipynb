```python
# int data types
a=10
print(a)
b=-10
print(b)
c=0
print(c)
print("\n")

# float data types
d= 10.5
print (d)
e= 10.5
print (e)
f= 0.5
print (f)
print("\n")

# complex data types
g= 10j
print (g)
h= -10j
print (h)
i= 0.10j
print (i)
print("\n")

#string data types
j="urooj_khan"
print(j)
print(j[7])
print("\n")

# types of data (a-c):
print(type(a))
print(type(b))
print(type(c))
print("\n")
# types of data (d-f):

print (type(d))
print (type(e))
print (type(f))
print("\n")
# types of data (g-i):

print (type(g))
print (type(h))
print (type(i))
print("\n")

# type of data j:
print(type(j))

```

```python
#if control flow
height=5.6
if height<=6:
  print(" not qualified")
  print("\n")

#if-else control flow
  age=18
  if age>=18:
    print( "eligible ")
  else:
      print("not eligible")
  print("\n")

#if-elif control flow
  marks=80
  if marks>=95:
      print("grade A+")
  elif marks >=90:
      print("grade A")
  elif marks>=85:
      print("grade B")
  else:
     print("do better in next time")
  print("\n")

# nested if-else control flow
  age=20
  right_age=18

  if age>=18:
    if right_age<=18:
        print("good ")
    else:
        print("not good")
  else:
    print("applied in next year")
```
