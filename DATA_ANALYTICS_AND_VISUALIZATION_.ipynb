```python
# Plotting settings
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# 1. Bar Chart: Top 10 Artists by Streams
top_10_artists = df.groupby('artist')['total_streams_billions'].sum().sort_values(ascending=False).head(10)
plt.figure()
sns.barplot(x=top_10_artists.values, y=top_10_artists.index, palette='viridis')
plt.title('Top 10 Artists by Total Streams (Billions)', fontsize=15)
plt.xlabel('Streams (in Billions)')
plt.ylabel('Artist Name')
plt.show()

```

```python
# 2. Bar Chart: Genre Distribution
plt.figure()
sns.countplot(y='primary_genre', data=df, order=df['primary_genre'].value_counts().index, palette='magma')
plt.title('Frequency of Genres in Top 100', fontsize=15)
plt.show()
```

```python
# 3. Line Chart: Average Streams over Years
streams_trend = df.groupby('release_year')['total_streams_billions'].mean()
plt.figure()
sns.lineplot(x=streams_trend.index, y=streams_trend.values, marker='o', color='red', linewidth=2)
plt.title('Evolution of Streams by Release Year', fontsize=15)
plt.xlabel('Year of Release')
plt.ylabel('Average Streams (Billions)')
plt.show()
```

```python
# 4. Histogram: BPM Distribution
plt.figure()
sns.histplot(df['bpm'], bins=15, kde=True, color='teal')
plt.title('Distribution of Song Tempos (BPM)', fontsize=15)
plt.xlabel('Beats Per Minute (BPM)')
plt.show()

```

```python
# 5. Boxplot: Danceability by Top 5 Genres
top_5_genres = df['primary_genre'].value_counts().head(5).index
df_top_genres = df[df['primary_genre'].isin(top_5_genres)]
plt.figure()
sns.boxplot(x='primary_genre', y='danceability', data=df_top_genres, palette='Set3')
plt.title('Danceability Variation Across Top Genres', fontsize=15)
plt.show()
```

```python
# 6. Heatmap: Feature Correlation
plt.figure(figsize=(10, 8))
corr_matrix = df[['total_streams_billions', 'bpm', 'danceability', 'energy', 'valence', 'acousticness']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap: Audio Features vs Streams', fontsize=15)
plt.show()
```

**Q1: Which artists and countries dominate the global streaming charts?**

**Method:** Hum artist aur artist_country ke mutabiq streams ko aggregate karenge.

**Logic:** Isse humein market share ka pata chalta hai.

```python
# Code to find dominant artists and countries
top_artist_streams = df.groupby('artist')['total_streams_billions'].sum().sort_values(ascending=False).head(5)
top_countries = df['artist_country'].value_counts().head(5)

print("Top 5 Artists by Total Streams:\n", top_artist_streams)
print("\nTop 5 Countries by Number of Hits:\n", top_countries)
```

**Find through first question:**

USA aur UK market par kabiz hain, aur Ed Sheeran/The Weeknd jaise artists global leaders hain.

**Q2: Is there a specific "sweet spot" for tempo (BPM) and danceability?**

**Method:** Hum bpm aur danceability ka mean aur median nikalenge.

**Logic:** Naye gaane banate waqt "Target Tempo" set karne mein madad milti hai.

```python
# Finding the 'Sweet Spot'
avg_bpm = df['bpm'].median()
avg_dance = df['danceability'].mean()

print(f"Ideal BPM (Median): {avg_bpm}")
print(f"Ideal Danceability (Mean): {avg_dance:.2f}")
```

**Find through second question**

 115-125 BPM aur 0.70+ danceability ek "Hit Song" ka standard formula nazar aata hai.

**Q3: How has the volume of streams changed for songs released in different decades?**

**Method:** Hum release_year ko decades mein group karenge.

**Logic:** Yeh dikhata hai ke kya purane gaane ("Classics") streaming mein naye gaano ka muqabla kar sakte hain.

```python
# Grouping by Decade
df['decade'] = (df['release_year'] // 10) * 10
decade_analysis = df.groupby('decade')['total_streams_billions'].sum()

plt.figure(figsize=(8,5))
decade_analysis.plot(kind='bar', color='orange')
plt.title('Total Streams by Decade')
plt.show()
```

**Find this through third question**

2010s ke decade mein sabse zyada streams hain, kyunke streaming platforms isi dor mein aam hue.

**Q4: Do explicit songs perform better than clean versions?**

**Method:** Hum explicit column ke mutabiq streams ka average compare karenge.

**Logic:** Content creators ko faisla karne mein madad milti hai ke kya "Explicit" tag audience ko rokta hai ya khenchta hai.



```python
# Explicit vs Clean Comparison
explicit_performance = df.groupby('explicit')['total_streams_billions'].mean()
print("Average Streams (Explicit vs Clean):\n", explicit_performance)
```

**Find this through fourth question**

Explicit gaano ke streams aksar barabar ya thore zyada hote hain, jo zahir karta hai ke modern audience ko lyrics ki "censorship" se farq nahi parta.

**Q5: Which musical features are most correlated with high streams?**

**Method:** Hum total_streams_billions ka correlation check karenge baki features ke saath.

**Logic:** Yeh batata hai ke "Vibe" (Energy, Acousticness) ka popularity se kitna gehra talluq hai.

```python
# Correlation with Success Metric
features_corr = df[['total_streams_billions', 'energy', 'acousticness', 'valence']].corr()['total_streams_billions']
print("Correlation with Streams:\n", features_corr)
```

**Find this through fifth question**

Aksar acousticness ka negative correlation hota hai, matlab "Electronic/Produced" music "Unplugged" music se behtar perform karta hai top charts mein.
